```python
import numpy as np
print("Numpy imported successfully");
```

```python
##creating arrays
```

```python
arr = np.array([10,20,30,40,50])
print(arr)
print(type(arr))
```

```python
##shape of array
```

```python
print("Shape:", arr.shape);
```

```python
##reshaping arrays
```

```python
reshaped = arr.reshape(1, 5)
print(reshaped)
```

```python
##Indexing
```

```python
print("First Element", arr[0])
print("Fourth Element", arr[3])
```

```python
##Slicing
```

```python
print(arr[1:4])
```

```python
##Statistical Operations
```

```python
print("Mean: ", arr.mean())
print("Median: ", np.median(arr))
print("Sum: ", arr.sum())
print("Max: ", arr.max())
print("Min: ", arr.min())

```
